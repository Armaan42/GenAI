### Difference between sort & Sorted

* __sort()__ - a method that only works on lists, and changes the list in place (modifies the original). It returns None.

* __sorted()__ - a built-in function that works on any iterable (lists, tuples, strings, etc.), and returns a new sorted list, leaving the original unchanged.

```python
nums = [4, 1, 3, 2]
nums.sort()
print(nums)   # [1, 2, 3, 4]  -> original list is changed
```

```python
nums = [4, 1, 3, 2]
result = sorted(nums)
print(result)   # [1, 2, 3, 4]  -> new list
print(nums)     # [4, 1, 3, 2]  -> original is untouched
```

Given matrix = [[1, 2, 3], [4, 5, 6], [7, 8, 9]], write code to calculate the sum of all numbers in the matrix.


```python
matrix = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]

total = 0

for row in matrix:
    for col in row:
        total += col

print(total)
```

Write a function that takes a list of numbers and returns a new list with running totals — e.g. [1, 2, 3, 4] → [1, 3, 6, 10].

```python
def running_totals(nums):
    result = []
    total = 0
    for num in nums:
        total += num
        result.append(total)
    return result

print(running_totals([1, 2, 3, 4]))   # [1, 3, 6, 10]
```

# __Tuples__

A tuple is like a list, but immutable - once you create it, you can't change, add, or remove items. You create one using parentheses ( ), with items separated by commas.

__A few key things about tuples:__

1. Ordered - items keep their position, just like a list.
2. Indexed - access items by position, starting at 0:

```python
coordinates = (10, 20)
colors = ("red", "green", "blue")
```

```python
print(coordinates)
print(colors)
```

* Immutable - you cannot modify a tuple after creating it:

```python
  colors[0] = "purple"   # ❌ This throws an error!
```

* Mixed types allowed - just like lists:

```python
mixed = (1, "hello", 3.5, True)
```

```python
print(mixed)
```

* Single-item tuples need a trailing comma, or Python won't treat them as a tuple:

```python
not_a_tuple = (5)     # this is just the number 5
a_tuple = (5,)         # this IS a tuple
```
