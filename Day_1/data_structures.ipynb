### Difference between sort & Sorted

* __sort()__ - a method that only works on lists, and changes the list in place (modifies the original). It returns None.

* __sorted()__ - a built-in function that works on any iterable (lists, tuples, strings, etc.), and returns a new sorted list, leaving the original unchanged.

```python
nums = [4, 1, 3, 2]
nums.sort()
print(nums)   # [1, 2, 3, 4]  -> original list is changed
```

```python
nums = [4, 1, 3, 2]
result = sorted(nums)
print(result)   # [1, 2, 3, 4]  -> new list
print(nums)     # [4, 1, 3, 2]  -> original is untouched
```

Write a function that takes a list of numbers and returns a new list with running totals — e.g. [1, 2, 3, 4] → [1, 3, 6, 10].

```python
def running_totals(nums):
    result = []
    total = 0
    for num in nums:
        total += num
        result.append(total)
    return result

print(running_totals([1, 2, 3, 4]))   # [1, 3, 6, 10]
```

# __Tuples__

A tuple is like a list, but immutable - once you create it, you can't change, add, or remove items. You create one using parentheses ( ), with items separated by commas.

__A few key things about tuples:__

1. Ordered - items keep their position, just like a list.
2. Indexed - access items by position, starting at 0:

```python
coordinates = (10, 20)
colors = ("red", "green", "blue")
```

```python
print(coordinates)
print(colors)
```

* Immutable - you cannot modify a tuple after creating it:

```python
  colors[0] = "purple"   # ❌ This throws an error!
```

* Mixed types allowed - just like lists:

```python
mixed = (1, "hello", 3.5, True)
```

```python
print(mixed)
```

* Single-item tuples need a trailing comma, or Python won't treat them as a tuple:

```python
not_a_tuple = (5)     # this is just the number 5
a_tuple = (5,)         # this IS a tuple
```

# __Sets__

A set in Python is an unordered collection of unique values. You create one using curly braces { } or the set() function.

```python
fruits = {"apple", "banana", "cherry"}
print(fruits)
```

A few key things about sets:

* __Unordered__ - items have no fixed position, so you can't access them by index (fruits[0] won't work).
* __No duplicates__ - sets automatically remove repeated values:

```python
nums = {1, 2, 2, 3, 3, 3}
print(nums)   # {1, 2, 3}
```

* __Mutable__ - you can add or remove items, but the items inside a set must be immutable (so no lists, but tuples are fine

```python
fruits.add("orange")
fruits
```

```python
fruits.remove("banana")
```

```python
fruits
```

* __Empty set__ - you must use set(), not {}, since {} creates an empty dictionary

```python
empty_set = set()
```

Why use a set?

* Automatically eliminates duplicates from data.
* Very fast for checking whether an item exists (in operator).
* Supports mathematical set operations like union, intersection, and difference

```python
a = {1, 2, 3}
b = {2, 3, 4}

a | b # union
```

```python
a & b # intersection
```

```python
a - b # difference
```

Since sets are unordered, looping through one won't guarantee a specific order

# __Dictionary__

A dictionary in Python stores data as key-value pairs, instead of just a sequence of values. You create one using curly braces { } with key: value pairs.

```python
student = {"name": "Alice", "age": 22, "major": "Computer Science"}
```

A few key things about dictionaries:

* __Key-value pairs__ - each item has a unique key that maps to a value

```python
student["name"]   # "Alice"
student["age"]    # 22
```

* __Keys must be unique and immutable__ - strings, numbers, or tuples work as keys; lists don't
* __Values can be anything__ - numbers, strings, lists, even other dictionaries.
* __Ordered (since Python 3.7+)__ - items keep the order they were inserted in.
* __Mutable__ - you can add, change, or remove key-value pairs:

```python
student["age"] = 23            # update a value
student["gpa"] = 3.8           # add a new key-value pair
del student["major"]           # remove a key-value pair
```

Common operations:

```python
student.keys()     # dict_keys(['name', 'age', 'gpa'])
student.values()   # dict_values(['Alice', 23, 3.8])
student.items()    # dict_items([('name', 'Alice'), ('age', 23), ('gpa', 3.8)])

"name" in student   # True — checks if a key exists
```

Looping through a dictionary:

```python
for key, value in student.items():
    print(key, "-", value)
```

Why use a dictionary?

* Fast lookups by key (much faster than searching a list for a value).
* Great for representing structured, labeled data - like a record, config, or JSON-like data.
* Lets you map one value to another (like a phonebook: name -> number).

Given prices = {"pen": 10, "notebook": 50, "eraser": 5}, write code to find the item with the highest price.

```python
prices = {"pen": 10, "notebook": 50, "eraser": 5}

highest_price = 0
highest_item = ""

for item in prices:
    if prices[item] > highest_price:
        highest_price = prices[item]
        highest_item = item

print(highest_item)
print(highest_price)
```
