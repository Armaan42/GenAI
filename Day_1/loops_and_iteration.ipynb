## __Loops & iteration__

### Loops

check whether a number is even  or odd

```python
x = int(input("Enter a number: "))

if x % 2 == 0:
    print(x, "is an even number")
else:
    print(x, "is an odd number")
```

Allow: greater than 18 & less than 50 years of age

```python
x = int(input("Enter your age"))

if x < 18 or x > 50:
    print("Sorry!! you are not allowed")
else:
    print("Welcome to the party")
```

program to provide discounts/offers

```python
shopping_amt = int(input("Input you billed amount"))

if shopping_amt >= 500:
    print("You have won a voucher")
else:
    print("welcome to the party")
```

## Iterations

iterate over a list of intergs

```python
list_1 = [10, 20, 30, 40, 50]

for i in list_1:
    print(i)
```

iterate over a string

```python
str_1 = "India"
for i in str_1:
    print(i)
```

iterate over a dictionary

```python
student_data = {1:["sam", 25], 2: ["sharma", 25 ], 3: ["Ravi", 23]}
for i in student_data:
    print(i)
```

```python
for i in student_data.values():
    print(i)
```

```python
for i in student_data.items():
    print(i)
```

## __Comprehensions__

```python
list_1 = ["Automobiles", "Honda", "Benz", "Maruti", "Kia"]

list_2 = []

for i in list_1:
    list_2.append(i)
    
print(list_2)
```

```python
list_1 = ["Automobiles", "Honda", "Benz", "Maruti", "Kia"]

list_2 = []

for i in list_1:
    list_2.append(len(i))
    
print(list_2)
```

```python
list_1 = ["Automobiles", "Honda", "Benz", "Maruti", "Kia"]

list_2 = [len(i) for i in list_1]
    
print(list_2)
```

```python
list_1 = ["Automobiles", "Honda", "Benz", "Maruti", "Kia"]

d = {i: len(i) for i in list_1}
d
```
